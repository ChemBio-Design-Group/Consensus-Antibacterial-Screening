**GENERATION OF BEMIS–MURCKO SCAFFOLDS**

Extraction of Bemis-Murcko scaffolds from the dataset for network-based chemical space analysis.

Compatibility: Google Colab

STEP-BY-STEP OVERVIEW

**STEP 1****: Install and import the required libraries.**

The `pandas `library is used for data manipulation and CSV file handling, whereas RDKit is used for molecular structure processing and Bemis–Murcko scaffold extraction through the `Chem` and `MurckoScaffold modules.`

**STEP 2**: Define the input and output files and molecular structure column. **texto en negrita**

Specify the input CSV file `(Curated_Dataset.csv)` containing the molecular dataset and define the output file `(Bemis_Murcko_Scaffolds.csv)` for storing the extracted scaffolds.

**STEP 3****: Load and validate the input dataset.**

Load the input CSV file into a pandas DataFrame and verify that the specified `SMILES` column is present before initiating the scaffold extraction procedure. If the required molecular structure column is absent, the script generates an error message and terminates the analysis to prevent processing of an incorrectly formatted dataset.

**STEP 4****: Extract Bemis–Murcko scaffolds from the molecular structures.**

Process each molecular structure individually by converting its SMILES representation into an RDKit molecular object using `Chem.MolFromSmiles()`. For successfully parsed structures, the corresponding Bemis–Murcko scaffold is extracted using `MurckoScaffold.GetScaffoldForMol()`. The resulting scaffold is subsequently converted into a SMILES representation using `Chem.MolToSmiles()` and incorporated into the original dataset as a new Scaffolds column.

**STEP 5****: Save the resulting scaffold dataset.**

Retain the original molecular information together with the corresponding Bemis–Murcko scaffold representation and export the resulting dataset as `Bemis_Murcko_Scaffolds.csv`. The exported file can subsequently be used for downstream structural analyses.


```python
# ============================================================
# STEP 1
# ============================================================
# 1. Install required libraries
# ============================================================
!pip install rdkit
```

```python
# ============================================================
# STEP 1
# ============================================================
# 2. Import required libraries
# ============================================================
import pandas as pd
from rdkit import Chem
from rdkit.Chem.Scaffolds import MurckoScaffold
```

```python
# ============================================================
# STEP 2
# ============================================================
# 3. Define input and output files
# ============================================================
input_file = "Curated_Dataset.csv"
output_file = "Bemis_Murcko_Scaffolds.csv"
```

```python
# ============================================================
# STEP 2
# ============================================================
# 4. Define the molecular structure column
# ============================================================
structure_column = "SMILES"
```

```python
# ============================================================
# STEP 3
# ============================================================
# 5. Load the input dataset
# ============================================================
df = pd.read_csv(input_file)
```

```python
# ============================================================
# STEP 3
# ============================================================
# 6. Validate the molecular structure column
# ============================================================
if structure_column not in df.columns:
    raise KeyError(
        f"The column '{structure_column}' is not present in the input CSV file."
    )
```

```python
# ============================================================
# STEP 4
# ============================================================
# 7. Define the Bemis–Murcko scaffold extraction function
# ============================================================
def get_bemis_murcko_scaffold(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles)

        if mol:
            # Extract the Bemis–Murcko scaffold
            scaffold = MurckoScaffold.GetScaffoldForMol(mol)

            # Convert the scaffold to SMILES representation
            return Chem.MolToSmiles(scaffold)

        else:
            # Return None for invalid molecular structures
            return None

    except Exception:
        # Return None if an error occurs during processing
        return None
```

```python
# ============================================================
# STEP 4
# ============================================================
# 8. Extract Bemis–Murcko scaffolds
# ============================================================
df["Scaffolds"] = df[structure_column].apply(get_bemis_murcko_scaffold)
```

```python
# ============================================================
# STEP 5
# ============================================================
# 9. Generate the final scaffold  and save the dataset
# ============================================================
# The original molecular information is retained,
# together with the corresponding Bemis–Murcko scaffold.

df.to_csv(output_file, index=False)

print(
    f"Dataset containing the extracted Bemis–Murcko scaffolds "
    f"was saved as: {output_file}"
)
```
